## Q1. Sensor readings

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add the average measurement error
corrected_temperature = temperature + 2
print("Corrected temperatures:", corrected_temperature)

# b. Celsius to Fahrenheit
fahrenheit = (9 / 5) * corrected_temperature + 32
print("Fahrenheit temperatures:", fahrenheit)

# c. Readings greater than 32°C
above_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", above_32)

# d. Number of readings greater than 32°C
count_above_32 = np.sum(corrected_temperature > 32)
print("Number of readings above 32°C:", count_above_32)
```

**e. Explanation:** Vectorization applies an operation to the whole NumPy array at once instead of processing each value separately with a loop. Boolean indexing creates a condition for the complete array and directly selects the values that satisfy it. NumPy performs these operations efficiently using optimized array operations.

## Q2. Daily steps

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Total steps
print("Total steps:", steps.sum())

# b. Mean number of steps
print("Mean steps:", steps.mean())

# c. Maximum and minimum
print("Maximum:", steps.max())
print("Minimum:", steps.min())

# d. Total steps for each day (axis=0)
print("Total for each day:", steps.sum(axis=0))

# e. Total steps for each user (axis=1)
print("Total for each user:", steps.sum(axis=1))

# f. Position of maximum value
max_position = np.unravel_index(steps.argmax(), steps.shape)
print("Position of maximum value:", max_position)
print("Maximum steps:", steps[max_position])
```

## Q3. NumPy arrays, slicing, copying and reshaping

```python
# a. Original array
original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

# b-c. Slice and modify it
subset = original[1:5]
subset[0] = 999
print("After changing subset:")
print("Original:", original)
print("Subset:", subset)

# d. Slice using copy()
copied = original[1:5].copy()
copied[0] = 500
print("\nAfter changing copied array:")
print("Original:", original)
print("Copied:", copied)

# e. Create 1 to 12 and reshape to 3 x 4
matrix = np.arange(1, 13).reshape(3, 4)
print("\n3 x 4 matrix:")
print(matrix)

# f. Indexing and slicing
print("\nFirst row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

# g. flatten() and ravel()
flat_array = matrix.flatten()
ravel_array = matrix.ravel()
print("\nUsing flatten():", flat_array)
print("Using ravel():", ravel_array)

# h. Change ravel() result
ravel_array[0] = 1000
print("\nAfter changing ravel():")
print("Ravel array:", ravel_array)
print("Original matrix:", matrix)

# i. Change flatten() result
flat_array[1] = 2000
print("\nAfter changing flatten():")
print("Flatten array:", flat_array)
print("Original matrix:", matrix)

# j. Array information
print("\nShape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

## Q4. Cognitive-assistive system and Ordinary Least Squares

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a. Shape and dimensions
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

# b. Transpose
XT = X.T
print("\nX transpose:")
print(XT)
print("\nThe transpose changes the rows into columns. Each row of XT corresponds to one feature from X.")

# c. X.T @ X
XTX = XT @ X
print("\nX.T @ X:")
print(XTX)

# d. Inverse
XTX_inv = np.linalg.inv(XTX)
print("\nInverse of X.T @ X:")
print(XTX_inv)

# e. Ordinary Least Squares: beta = (X.T X)^-1 X.T y
beta = XTX_inv @ XT @ y
print("\nOLS coefficients:")
print(beta)

# f. Meaning of coefficients
print("\nCoefficient interpretation:")
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])
print("Each coefficient shows the estimated change in assistance score for a one-unit change in that feature, while the other features are kept fixed.")

# g. Prediction for a new user
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("\nPredicted assistance score:", predicted_score)
```
