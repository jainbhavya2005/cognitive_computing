Q1. Write a unique paragraph (5-6 sentences) about your favorite topic (e.g., sports,
technology, food, books, etc.).
1. Convert text to lowercase and remove punctuaƟon using re.
2. Tokenize the text into words and sentences.
3. Split using split() and word_tokenize() and compare how Python split and NLTK’s
word_tokenize() differ.
4. Remove stopwords (using NLTK's stopwords list).
5. Display word frequency distribuƟon (excluding stopwords).

```python
import re
import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = """Cybersecurity is an essential part of our digital world because it protects people and organizations from cyber crimes. With the increasing use of the internet, threats such as hacking, phishing, identity theft, malware, and online fraud are becoming more common. Strong passwords, two-factor authentication, antivirus software, and data encryption help protect sensitive information from unauthorized access. People should avoid clicking on suspicious links, sharing personal information on unknown websites, and downloading files from untrusted sources. Organizations should regularly update their security systems and educate employees about cyber threats to prevent data breaches. By following proper cybersecurity practices and staying aware of online risks, we can create a safer digital environment and protect ourselves from cyber crimes."""

clean_text = re.sub(r'[^\w\s]', '', text.lower())

print("1. Cleaned Text:")
print(clean_text)

words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("\n2. Word Tokens:")
print(words)

print("\nSentence Tokens:")
print(sentences)

print("\n3. Python split():")
print(clean_text.split())

print("\nNLTK word_tokenize():")
print(word_tokenize(text.lower()))

stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word not in stop_words]

print("\n4. Words after Stopword Removal:")
print(filtered_words)

fdist = FreqDist(filtered_words)

print("\n5. Word Frequency Distribution:")
print(fdist.most_common())

fdist.plot(10, title="Top 10 Most Frequent Words in Cybersecurity Text")
plt.show()
```

Q2. Using the same paragraph from Q1:
1. Extract all words with only alphabets using re.findall()
2. Remove stop words using NLTK’s stopword list
3. Perform stemming with PorterStemmer
4. Perform lemmaƟzaƟon with WordNetLemmaƟzer
5. Compare the stemmed and lemmaƟzed outputs and explain when you’d prefer one over
the other.

```python
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

text = """Cybersecurity is an essential part of our digital world because it protects people and organizations from cyber crimes. With the increasing use of the internet, threats such as hacking, phishing, identity theft, malware, and online fraud are becoming more common. Strong passwords, two-factor authentication, antivirus software, and data encryption help protect sensitive information from unauthorized access. People should avoid clicking on suspicious links, sharing personal information on unknown websites, and downloading files from untrusted sources. Organizations should regularly update their security systems and educate employees about cyber threats to prevent data breaches. By following proper cybersecurity practices and staying aware of online risks, we can create a safer digital environment and protect ourselves from cyber crimes."""

words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

print("1. Alphabetic Words:")
print(words)

stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word not in stop_words]

print("\n2. After Stopword Removal:")
print(filtered_words)

ps = PorterStemmer()

stemmed_words = [ps.stem(word) for word in filtered_words]

print("\n3. Stemmed Words:")
print(stemmed_words)

lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("\n4. Lemmatized Words:")
print(lemmatized_words)

print("\n5. Comparison:")

for original, stem, lemma in zip(filtered_words, stemmed_words, lemmatized_words):
    print(original, " -> Stemming:", stem, "| Lemmatization:", lemma)
```

Q3. Choose 3 short texts of your own (e.g., different news headlines, product reviews).
1. Use CountVectorizer to generate the Bag of Words representaƟon.
2. Use TfidfVectorizer to compute TF-IDF scores.
3. Print and interpret the top 3 keywords from each text using TF-IDF.

```python
import nltk
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

texts = [
    """ViTrace is an advanced face recognition system that can identify a person from facial images. It uses computer vision and deep learning to analyze facial features and match faces with available records. The system helps locate a person using face based identification. It provides an efficient approach to automated person identification.""",

    """ViTrace can track a person across camera recordings and video clips using facial recognition technology. It analyzes video frames to detect and follow a particular person throughout the footage. The system helps monitor movement and identify appearances across different timestamps. It can be useful for authorized security monitoring and investigation.""",

    """ViTrace is designed to improve security through intelligent face detection, person identification, and video tracking. It combines artificial intelligence, computer vision, and machine learning to analyze visual data. The system can search for a target person in available camera footage and recognize matching appearances. ViTrace aims to support authorized investigations, surveillance, and security operations while maintaining privacy and responsible data usage."""
]

cv = CountVectorizer()

bow_matrix = cv.fit_transform(texts)

print("1. BAG OF WORDS REPRESENTATION")

print("\nVocabulary:")
print(cv.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())

tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(texts)

print("\n\n2. TF-IDF SCORES")

print("\nVocabulary:")
print(tfidf.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(np.round(tfidf_matrix.toarray(), 3))

feature_names = tfidf.get_feature_names_out()

print("\n\n3. TOP 3 KEYWORDS FROM EACH TEXT")

for i in range(len(texts)):

    scores = tfidf_matrix[i].toarray().flatten()

    top_indices = scores.argsort()[-3:][::-1]

    print("\nText", i + 1)

    for index in top_indices:
        print(feature_names[index], ":", round(scores[index], 3))
```

Q4. Write 2 short texts (4–6 lines each) describing two different technologies (e.g., AI vs
Blockchain).
1. Preprocess and tokenize both texts.
2. Calculate:
a. Jaccard Similarity using sets
b. Cosine Similarity using TfidfVectorizer + cosine_similarity()
c. Analyze which similarity metric gives beƩer insights in your case.

```python

import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text1 = """
Artificial Intelligence uses machine learning and computer vision
to recognize human faces and identify people in images.
ViTrace uses AI technology to detect faces and track people
across camera recordings and video clips.
It helps improve security and surveillance through automation.
"""

text2 = """
Blockchain is a decentralized technology used to store information
securely across multiple computer systems.
It uses cryptography to protect digital records from modification.
Blockchain improves data security, transparency, and trust.
It is widely used in digital transactions and secure data management.
"""

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    words = text.split()
    return words

words1 = preprocess(text1)
words2 = preprocess(text2)

print("Text 1 Tokens:")
print(words1)

print("\nText 2 Tokens:")
print(words2)

set1 = set(words1)
set2 = set(words2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard = len(intersection) / len(union)

print("\nJaccard Similarity:", round(jaccard, 4))
print("Common Words:", intersection)

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform([text1, text2])

cosine = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

print("\nCosine Similarity:", round(cosine[0][0], 4))

print("\nAnalysis:")

if cosine[0][0] > jaccard:
    print("Cosine similarity gives better insight because")
    print("it considers the importance of words using TF-IDF.")
else:
    print("Jaccard similarity gives a better indication of")
    print("the overlap between unique words.")
```

Q5. Write a short review for a product or service.
1. Use TextBlob or VADER to find polarity & subjecƟvity for each review.
2. Classify reviews into PosiƟve / NegaƟve / Neutral.
3. Create a word cloud using the wordcloud library for all posiƟve reviews.

```python
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt

reviews = [
    "ViTrace is an excellent face recognition system. "
    "It accurately identifies people and provides fast results. "
    "The tracking feature is very useful for security.",

    "ViTrace is a powerful AI based surveillance system. "
    "It can detect faces and track people across video clips. "
    "The technology is impressive and helpful.",

    "ViTrace sometimes gives incorrect face matches. "
    "The tracking process can be slow in some videos. "
    "The system needs improvement in accuracy and performance."
]

positive_reviews = []

print("SENTIMENT ANALYSIS OF VITRACE REVIEWS\n")

for i, review in enumerate(reviews):

    analysis = TextBlob(review)

    polarity = analysis.sentiment.polarity
    subjectivity = analysis.sentiment.subjectivity

    if polarity > 0:
        sentiment = "Positive"
        positive_reviews.append(review)

    elif polarity < 0:
        sentiment = "Negative"

    else:
        sentiment = "Neutral"

    print("Review", i + 1)
    print("Text:", review)
    print("Polarity:", round(polarity, 3))
    print("Subjectivity:", round(subjectivity, 3))
    print("Sentiment:", sentiment)
    print("--------------------------------")

if positive_reviews:

    positive_text = " ".join(positive_reviews)

    wordcloud = WordCloud(
        width=1000,
        height=500,
        background_color="white",
        colormap="viridis"
    ).generate(positive_text)

    plt.figure(figsize=(12, 6))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Positive ViTrace Product Reviews")
    plt.show()
```

Q6. Choose your own paragraph (~100 words) as training data.
1. Tokenize text using Tokenizer() from keras.preprocessing.text
2. Create input sequences and build a simple LSTM or Dense model
3. Train the model and generate 2–3 new lines of text starƟng from any seed word you
provide.

```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense


text = """
ViTrace is an intelligent artificial intelligence based security system.
It uses computer vision and deep learning to detect and recognize human faces.
The system can identify a person using facial images and available records.
ViTrace also tracks people across camera recordings and video clips.
It analyzes video frames to detect faces and monitor human movement.
The system helps improve security and surveillance through automation.
Face recognition makes identification faster and more efficient.
Machine learning helps the system analyze visual information.
ViTrace combines artificial intelligence computer vision and machine learning.
It can support authorized investigations and security monitoring.
The main goal of ViTrace is to provide intelligent person identification
and tracking while maintaining privacy and responsible data usage.
"""

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Total Unique Words:", total_words)
input_sequences = []

for line in text.split("."):
    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i + 1]
        input_sequences.append(n_gram_sequence)

max_sequence_len = max(len(seq) for seq in input_sequences)

input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding='pre'
)

X = input_sequences[:, :-1]
y = input_sequences[:, -1]

print("Input Shape:", X.shape)
print("Output Shape:", y.shape)

model = Sequential()

model.add(Embedding(
    total_words,
    64,
    input_length=max_sequence_len - 1
))

model.add(LSTM(100))

model.add(Dense(total_words, activation='softmax'))

model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.summary()

model.fit(
    X,
    y,
    epochs=100,
    verbose=1
)

def generate_text(seed_text, next_words):

    for _ in range(next_words):

        token_list = tokenizer.texts_to_sequences([seed_text])[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding='pre'
        )

        predicted = model.predict(token_list, verbose=0)

        predicted_word_index = np.argmax(predicted, axis=1)[0]

        output_word = ""

        for word, index in tokenizer.word_index.items():
            if index == predicted_word_index:
                output_word = word
                break

        if output_word == "":
            break

        seed_text += " " + output_word

    return seed_text

print("\nGENERATED TEXT:\n")

for i in range(3):

    generated = generate_text("ViTrace", 10)

    print("Line", i + 1, ":", generated)
```
