## Q1. Lists

```python
roll_number = input("Enter your roll number: ")
digits = [int(digit) * 10 for digit in roll_number if digit.isdigit()]
L = digits.copy()
print(L)

L.append(250)
print(L)

L.insert(2, 350)
print(L)

if L:
    L.remove(L[0])
print(L)

if L:
    L.pop()
print(L)

L.sort()
print(L)

L.sort(reverse=True)
print(L)

print(L[:3])
print(L[-3:])

average = sum(L) / len(L) if L else 0
above_average = [value for value in L if value > average]
print(above_average)
```

## Q2. Tuples

```python
scores = tuple(L[:8])
print("Scores:", scores)

if scores:
    highest = max(scores)
    lowest = min(scores)
    print("Highest score:", highest)
    print("Highest score index:", scores.index(highest))
    print("Lowest score:", lowest)
    print("Lowest score count:", scores.count(lowest))

print("Reversed list:", list(reversed(scores)))

try:
    scores[0] = 100
except TypeError as error:
    print(error)
    print("Tuples are immutable, so their elements cannot be changed after creation.")

score = int(input("Enter a score: "))
if score in scores:
    print("First occurrence index:", scores.index(score))
else:
    print("Score not present")

if len(scores) >= 2:
    first_score, second_score, *remaining_scores = scores
    print(first_score, second_score, remaining_scores)
```

## Q3. Random numbers

```python
import random
from collections import Counter

random.seed(int(roll_number))
numbers = [random.randint(100, 900) for _ in range(100)]

def is_prime(number):
    if number < 2:
        return False
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False
    return True

odd_numbers = [number for number in numbers if number % 2 != 0]
even_numbers = [number for number in numbers if number % 2 == 0]
prime_numbers = [number for number in numbers if is_prime(number)]

print("Numbers:", numbers)
print("Odd count:", len(odd_numbers))
print("Even count:", len(even_numbers))
print("Prime count:", len(prime_numbers))
print("Prime numbers:", prime_numbers)

counts = Counter(numbers)
most_common_number, frequency = counts.most_common(1)[0]
print("Most frequent number:", most_common_number)
print("Frequency:", frequency)
```

## Q4. Sets

```python
original_digits = [int(digit) for digit in roll_number if digit.isdigit()]
A = {digit * 7 for digit in original_digits}
B = {digit * 9 for digit in original_digits}

print("A:", A)
print("B:", B)
print("Union:", A.union(B))
print("Intersection:", A.intersection(B))
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
print("Symmetric difference:", A.symmetric_difference(B))
print("A subset of B:", A.issubset(B))
print("B superset of A:", B.issuperset(A))
print("difference() gives values in one set but not the other; symmetric_difference() combines both one-sided differences.")

X = int(input("Enter a value to discard from A: "))
A.discard(X)
print("A after discard:", A)
print("discard() does not raise an error if the value is missing, unlike remove().")
```

## Q5. Dictionaries

```python
my_dict = {
    "name": input("Enter your name: "),
    "roll_no": roll_number,
    "branch": input("Enter your branch: "),
    "age": int(input("Enter your age: ")),
    "city": input("Enter your home city: ")
}

my_dict["location"] = my_dict.pop("city")
my_dict["cgpa"] = float(input("Enter your CGPA: "))
my_dict["age"] += 1

print(my_dict)

dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")
print(dict_pop)

dict_del = my_dict.copy()
del dict_del["branch"]
print(dict_del)
print("pop() returns the removed value, while del removes the key without returning it.")

for key, value in my_dict.items():
    print(f"{key} → {value}")

if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email is not available")

friend_dict = {
    "name": "Aman",
    "roll_no": "12345678",
    "branch": "CSE",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}
print(merged_dict)
print("When keys overlap, the values from the dictionary written last take priority.")

string_values = {key: value for key, value in my_dict.items() if isinstance(value, str)}
print(string_values)
```
